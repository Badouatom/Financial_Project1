```python
returns = pd.DataFrame({
    "Asset_A": [0.04, -0.02, 0.05, 0.01, -0.03, 0.06],
    "Asset_B": [0.02, 0.01, 0.03, -0.01, -0.02, 0.02],
    "Asset_C": [-0.01, 0.03, -0.02, 0.02, 0.04, -0.01],
}, index=["Jan", "Feb", "Mar", "Apr", "May", "Jun"])
```

```python
returns
```

```python
#  return mean by Asset
mean_returns=returns.mean()
print(mean_returns)
# return of high Asset mean
print("\nHighest Mean Return:", mean_returns.idxmax())
```

```python
# Why is the mean important in financial analysis?
# The mean helps us answer three questions:
# 1. Performance: Which asset generated the highest average return?
# 2. Comparison: How do the average returns of different investments compare?
# 3. Portfolio management: What return might we use as an estimate when constructing a portfolio?
```

```python
# Calculate sample variance
sample_variance = returns.var(ddof=1)
print(sample_variance)
```

```python
# 3. Why variance matters in quantitative finance
# Variance helps analysts evaluate how much uncertainty exists around an asset's average return.
# In practical terms:
# - Higher variance: Returns fluctuate more widely, indicating greater variability and uncertainty.
# - Lower variance: Returns are more concentrated around their mean, indicating greater historical stability.
```

```python
# Calculate the Sampale Standard deviation
sample_std = returns.std(ddof=1)

print(sample_std)
```

```python
# Return by volatity rank

volatility_ranking = returns.std(ddof=1).sort_values(
    ascending=False
)

print(volatility_ranking)
```

```python
# Ranking from highest to lowest volatility:
# 1. Asset A — Most volatile (3.76%)
# 2. Asset C — Moderate volatility (2.48%)
# 3. Asset B — Least volatile (1.94%)
```

```python
# mean, variance, standard Deviation  recap
recp = pd.DataFrame({
    "Return Means": returns.mean(),
    "Return Variance": returns.var(ddof=1),
    "Return Volatility": returns.std(ddof=1).map("{:.2%}".format) #  map("{:.2%}".format) for the percentage. 
})

recp.index.name = "Asset"

print(recp.round(6))
```

```python
#Analyze results:
# Asset A: Has the highest historical average return, but also the highest volatility. Its monthly returns fluctuate more substantially.
# Asset B: Has the lowest volatility, meaning its monthly returns were relatively more stable.
# Asset C: Has the same mean return as Asset B but higher volatility. Based only on these two statistics, Asset B has the more favorable risk-return profile.
```
